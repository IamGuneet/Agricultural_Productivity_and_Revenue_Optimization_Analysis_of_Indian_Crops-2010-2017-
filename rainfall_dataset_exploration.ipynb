```python
import pandas as pd
```

```python
df = pd.read_csv('rainfall_data.csv')
```

```python
df['YEAR'].unique
```

```python
state_mapping = {
    'Chhattisgarh': ['Chhattisgarh'],
    
    'Madhya Pradesh': [
        'West Madhya Pradesh',
        'East Madhya Pradesh'
    ],
    
    'Andhra Pradesh': [
        'Coastal Andhra Pradesh',
        'Rayalseema'
    ],
    
    'Telangana': ['Telangana'],
    
    'Karnataka': [
        'Coastal Karnataka',
        'North Interior Karnataka',
        'South Interior Karnataka'
    ],
    
    'Tamil Nadu': ['Tamil Nadu'],
    
    'Maharashtra': [
        'Konkan & Goa',
        'Madhya Maharashtra',
        'Matathwada',
        'Vidarbha'
    ],
    
    'Gujarat': [
        'Gujarat Region',
        'Saurashtra & Kutch'
    ],
    
    'Rajasthan': [
        'West Rajasthan',
        'East Rajasthan'
    ],
    
    'Punjab': ['Punjab'],
    
    'Haryana': [
        'Haryana Delhi & Chandigarh'
    ],
    
    'Uttar Pradesh': [
        'East Uttar Pradesh',
        'West Uttar Pradesh'
    ],
    
    'Uttarakhand': ['Uttarakhand'],
    
    'Assam': [
        'Assam & Meghalaya'
    ],
    
    'Himachal Pradesh': [
        'Himachal Pradesh'
    ],
    
    'Kerala': ['Kerala'],
    
    'Orissa': ['Orissa'],
    
    'West Bengal': [
        'Sub Himalayan West Bengal & Sikkim',
        'Gangetic West Bengal'
    ],
    
    'Bihar': ['Bihar'],
    
    'Jharkhand': ['Jharkhand']
}
```

```python
target_subdivisions = [
    subdivision
    for sublist in state_mapping.values()
    for subdivision in sublist
]
```

```python
target_subdivisions
```

```python
filtered_rainfall_df = df[
    (df['SUBDIVISION'].isin(target_subdivisions)) &
    (df['YEAR'].between(2010, 2017))
]
```

```python
filtered_rainfall_df
```

```python
reverse_mapping = {}

for state, subdivisions in state_mapping.items():
    for subdivision in subdivisions:
        reverse_mapping[subdivision] = state

filtered_rainfall_df['State_Name'] = (
    filtered_rainfall_df['SUBDIVISION']
    .map(reverse_mapping)
)
```

```python
filtered_rainfall_df
```

```python
filtered_rainfall_df.to_csv('filtered_rainfall.csv',index=False)
```
