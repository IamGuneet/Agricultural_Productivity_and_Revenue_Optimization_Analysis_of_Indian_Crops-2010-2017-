```python
import re

cols = df.columns

crop_columns = [col for col in cols if re.search(r'AREA',col)]
len(crop_columns)
crop_names = [re.search(r'^(.*?)\s*AREA', col).group(1) for col in crop_columns]
df.columns
```

Keeping only high value crops

```python
high_value_crops = [
    'RICE',
    'WHEAT',
    'GROUNDNUT',
    'SESAMUM',
    'RAPESEED AND MUSTARD',
    'SUNFLOWER',
    'SAFFLOWER',
    'SOYABEAN',
    'OILSEEDS',
    'SUGARCANE',
    'COTTON',
]

base_cols = [
    'Dist Code',
    'Year',
    'State Code',
    'State Name',
    'Dist Name'
]

crop_cols = [
    col for col in df.columns
    if any(crop in col for crop in high_value_crops)
]

ndf = df[base_cols+crop_cols]
```

```python
ndf
```

```python
ndf.isnull().sum()
```

```python
zero_counts = (ndf == 0).sum()
total_len = len(ndf)
zero_df = pd.DataFrame(zero_counts)
zero_df['total_len'] = total_len
zero_df.rename(columns={
    0:"Count of 0"
},inplace=True)
```

```python
zero_df['Percentage of 0 values'] = ((zero_df['Count of 0']/zero_df['total_len']) * 100).round(2)
zero_df 
```

```python
ndf.to_csv('filtered_crops.csv',index=False)
```
